# Experiment 8(b): Extract Marriage Relationships

**Aim:** Use spaCy's NLP pipeline and pattern matching to identify marriage relationships.

Named-entity recognition can miss names depending on the model. The pattern is designed for the manual's simple example format.

````python
# Install if needed in a notebook cell: %pip install spacy
# Then download the English model in a separate cell if needed:
# !python -m spacy download en_core_web_sm

import spacy
from spacy.matcher import Matcher

nlp = spacy.load("en_core_web_sm")
matcher = Matcher(nlp.vocab)
pattern = [
    {"ENT_TYPE": "PERSON"},
    {"LOWER": {"IN": ["and", "&"]}},
    {"ENT_TYPE": "PERSON"},
    {"LOWER": {"IN": ["married", "wed"]}}
]
matcher.add("MARRIAGE", [pattern])

examples = [
    "Mark and Emily married yesterday.",
    "John and Mary married last year."
]
for text in examples:
    doc = nlp(text)
    matches = matcher(doc)
    found = False
    for match_id, start, end in matches:
        span = doc[start:end]
        people = [ent.text for ent in doc.ents if ent.label_ == "PERSON" and ent.start >= start and ent.end <= end]
        if len(people) >= 2:
            print(f"{people[0]} —[spouse]→ {people[1]}")
            found = True
    if not found:
        print("No marriage relationship detected:", text)
````

John —[spouse]→ Mary
