# Experiment 1(b): Sentence Tokenization

**Aim:** Split text into sentences using NLTK.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

sent_tokenize() creates sentence-level tokens.

````python
import nltk
from nltk.tokenize import sent_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

text = "Hello everyone. Welcome to class. You are studying NLP article"
print(sent_tokenize(text))
````

## Expected output

```text
['Hello everyone.', 'Welcome to class.', 'You are studying NLP article']
```
