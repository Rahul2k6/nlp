# Experiment 3: Tokenization and POS Tagging (Morphological Analysis)

**Aim:** Tokenize sentences and assign Part-of-Speech tags.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

POS tags can vary slightly by NLTK version.

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

for package in ["punkt", "punkt_tab", "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]:
    nltk.download(package, quiet=True)

def morphological_analysis(sentence):
    return pos_tag(word_tokenize(sentence))

sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "The scientists who discovered the new element were awarded the Nobel Prize for their groundbreaking research.",
    "She sells seashells by the seashore."
]
for sentence in sentences:
    print(f"for the sentence --> {sentence}")
    print("-----------------------")
    for token, tag in morphological_analysis(sentence):
        print(f"{token}: {tag}")
````
