# Experiment 5(b): Lemmatize Words Using WordNet

**Aim:** Show how WordNet lemmatization depends on the POS tag.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

POS values: n=noun, v=verb, a=adjective.

````python
import nltk
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words = ["running", "flies", "leaves", "better", "studying"]

print(f"{'Word':<12}{'Noun':<16}{'Verb':<16}{'Adjective':<16}")
print("-" * 60)
for word in words:
    print(f"{word:<12}{lemmatizer.lemmatize(word, pos='n'):<16}{lemmatizer.lemmatize(word, pos='v'):<16}{lemmatizer.lemmatize(word, pos='a'):<16}")
````

## Expected output

```text
Word        Noun            Verb            Adjective
------------------------------------------------------------
running     running         run             running
flies       fly             fly             flies
leaves      leaf            leave           leaves
better      better          better          good
studying    studying        study           studying
```
