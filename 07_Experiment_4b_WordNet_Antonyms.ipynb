# Experiment 4(b): Get Antonyms from WordNet

**Aim:** Retrieve unique antonyms associated with the word good.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

WordNet returns antonyms only where a lexical antonym is recorded.

````python
import nltk
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.corpus import wordnet

antonyms = []
for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        for antonym in lemma.antonyms():
            antonyms.append(antonym.name())
print(set(antonyms))
````

## Expected output

```text
{'badness', 'evilness', 'bad', 'evil', 'ill'}
```
