# Experiment 2(b): Count Word Frequency in a File

**Aim:** Count word frequencies from a text file.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

Example test.txt:
Mango banana apple pear
Banana grapes strawberry
Apple pear mango banana
Kiwi apple mango strawberry

````python
import re
from collections import Counter

# Create test.txt in the same folder before running.
with open("test.txt", encoding="utf-8") as f:
    text = f.read().lower()
words = re.findall(r'\w+', text)
freq = Counter(words)

for word, count in freq.most_common():
    print(word, count)
````

## Expected output

```text
mango 3
banana 3
apple 3
pear 2
strawberry 2
grapes 1
kiwi 1
```
