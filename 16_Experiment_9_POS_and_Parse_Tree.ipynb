A context-free grammar (CFG) defines valid sentence structures; ChartParser finds parses allowed by the grammar.

````python
import nltk
from nltk import CFG
from nltk.tree import Tree

grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N | Det N PP
VP -> V PP
PP -> P NP
Det -> 'the'
N -> 'cat' | 'mat'
V -> 'sat'
P -> 'on'
""")
parser = nltk.ChartParser(grammar)
sentence = "the cat sat on the mat".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
````

```text
(S (NP (Det the) (N cat) (VP ...))  # tree structure depends on grammar
```
