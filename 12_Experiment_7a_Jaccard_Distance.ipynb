# Experiment 7(a): Jaccard Distance Between Strings

**Aim:** Compute Jaccard distance between character-bigram sets of word pairs.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

Jaccard distance = 1 - (intersection size / union size). The manual uses NLTK n-grams.

````python
from nltk.util import ngrams

def jaccard_distance(word1, word2, n=2):
    set1 = set(ngrams(word1, n))
    set2 = set(ngrams(word2, n))
    if not set1 and not set2:
        return 0.0
    return 1 - len(set1 & set2) / len(set1 | set2)

pairs = [("night", "nacht"), ("book", "back"), ("cat", "cat")]
for a, b in pairs:
    print(f"Jaccard distance between {a} and {b}: {jaccard_distance(a, b):.3f}")
````

Jaccard distance between cat and cat: 0.000
