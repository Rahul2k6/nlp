# Experiment 2(a): Remove Numbers, Punctuation and Whitespace

**Aim:** Read input.txt, retain alphabetic characters, and write the result to output.txt.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

Example input.txt:
Hello , World! This is a test file.
123!^$% Hope you 4are doing well.
Provide me new feature34s of a te4&xt file.
End of file.

output.txt contains:
HelloWorldThisisatestfileHopeyouaredoingwellProvidemenewfeaturesofatextfileEndoffile

````python
def clean_text(text):
    return ''.join(char for char in text if char.isalpha())

def clean_file(input_file, output_file):
    try:
        with open(input_file, "r", encoding="utf-8") as file:
            content = file.read()
        cleaned_content = clean_text(content)
        with open(output_file, "w", encoding="utf-8") as file:
            file.write(cleaned_content)
        print(f"Cleaned content has been written to {output_file}.")
    except FileNotFoundError:
        print(f"Error: The file {input_file} was not found.")

# Create input.txt in the same folder before running.
clean_file("input.txt", "output.txt")
````

## Expected output

```text
Cleaned content has been written to output.txt.
```
