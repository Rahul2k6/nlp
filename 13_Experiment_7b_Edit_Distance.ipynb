# Experiment 7(b): Edit Distance Between Strings

**Aim:** Calculate the minimum insertions, deletions and substitutions needed to transform one word into another.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

Example input shown: kitten and sitting. Edit distance is the minimum number of single-character edits.

````python
from nltk.metrics import edit_distance

word1 = input("Enter first word: ")
word2 = input("Enter second word: ")
print("Edit distance:", edit_distance(word1, word2))
````

## Expected output

```text
Enter first word: kitten
Enter second word: sitting
Edit distance: 3
```
