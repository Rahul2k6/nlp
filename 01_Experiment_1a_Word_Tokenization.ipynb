# Experiment 1(a): Word Tokenization

**Aim:** Tokenize text into words and punctuation using NLTK.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

word_tokenize() splits text into word-level tokens and punctuation.

````python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")  # needed by some newer NLTK versions

text = "Hello everyone. Welcome to Web and Text Analytic class"
print(word_tokenize(text))
````

## Expected output

```text
['Hello', 'everyone', '.', 'Welcome', 'to', 'Web', 'and', 'Text', 'Analytic', 'class']
```
