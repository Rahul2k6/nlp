# Experiment 4(a): Get Synonyms from WordNet

**Aim:** Retrieve unique synonyms associated with the word good.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

Output set and order depend on WordNet data version.

````python
import nltk
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.corpus import wordnet

synonyms = []
for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())
print(set(synonyms))
````

## Expected output

```text
{'beneficial', 'effective', 'good', 'honest', 'well', 'skillful', ...}
```
