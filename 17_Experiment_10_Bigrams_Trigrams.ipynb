# Experiment 10: Print Bigrams and Trigrams

**Aim:** Generate bigrams and trigrams from a given text.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

A bigram has 2 consecutive tokens; a trigram has 3.

````python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
import nltk

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

text = "Natural language processing is very interesting"
tokens = word_tokenize(text.lower())

bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print("Bigrams:")
for item in bigrams:
    print(item)
print("\nTrigrams:")
for item in trigrams:
    print(item)
````

## Expected output

```text
Bigrams:
('natural', 'language')
('language', 'processing')
('processing', 'is')
('is', 'very')
('very', 'interesting')

Trigrams:
('natural', 'language', 'processing')
('language', 'processing', 'is')
('processing', 'is', 'very')
('is', 'very', 'interesting')
```
