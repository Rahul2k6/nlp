# Experiment 5(a): Compare Stemming and Lemmatization

**Aim:** Compare Porter stemming with WordNet noun lemmatization.

**Lab manual:** Vardhaman College NLP Lab (A8708, R22).

Run the setup/download cell if required. Expected outputs may vary slightly with NLTK/spaCy dataset or model versions.

The manual's comparison uses noun POS for lemmatization.

````python
import nltk
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
words = ["running", "flies", "leaves", "better", "studying"]

print(f"{'Word':<12}{'Stemmed':<12}{'Lemmatized':<12}")
print("-" * 36)
for word in words:
    print(f"{word:<12}{stemmer.stem(word):<12}{lemmatizer.lemmatize(word, pos='n'):<12}")
````

## Expected output

```text
Word        Stemmed     Lemmatized
------------------------------------
running     run         running
flies       fli         fly
leaves      leav        leaf
better      better      better
studying    studi       studying
```
